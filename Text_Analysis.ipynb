# Text Analysis

## Objective

To analyze a given text using Python and calculate the number of words, characters, vowels, and repeated words.

```python
text = """
Artificial intelligence is becoming an important part of modern technology and everyday life. 
Artificial intelligence allows computers to perform tasks that normally require human intelligence, 
such as understanding language, recognizing images, analyzing data, and making decisions. 
Python is one of the most popular programming languages used in artificial intelligence and data analysis 
because Python is simple to learn and provides many useful libraries. Data analysis helps organizations 
understand large amounts of information and make better decisions. Machine learning is another important 
area of artificial intelligence where computers learn patterns from data and improve their performance 
over time. Natural language processing helps computers understand and process human language. 
Natural language processing is used in applications such as chatbots, virtual assistants, translation 
systems, and text analysis tools. Learning Python, data analysis, machine learning, and natural language 
processing can help students develop useful technical skills and prepare for careers in the field of 
artificial intelligence.
"""
```

```python
words = text.split()

word_count = len(words)
character_count = len(text)

print("Number of Words      :", word_count)
print("Number of Characters :", character_count)
```

```python
vowels = "aeiou"
vowel_count = 0

for character in text.lower():
    if character in vowels:
        vowel_count += 1

print("Number of Vowels :", vowel_count)
```

```python
from collections import Counter

word_counts = Counter(word.lower().strip(".,!?") for word in words)

repeated_words = {}

for word, count in word_counts.items():
    if count > 1:
        repeated_words[word] = count

print("Repeated Words:")
for word, count in repeated_words.items():
    print(word, ":", count)
```

```python
print("Text Analysis")
print("-------------")
print("Number of Words      :", word_count)
print("Number of Characters :", character_count)
print("Number of Vowels     :", vowel_count)
print("Repeated Words       :", repeated_words)
```

## Explanation of Results

The given paragraph was analyzed using Python to identify basic text statistics.

The total number of words and characters was calculated using `split()` and `len()`.

The number of vowels was counted by checking each character in the text.

Repeated words were identified using Python's `Counter` and displayed along with their frequencies.

This analysis demonstrates basic text processing techniques using Python.
